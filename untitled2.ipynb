```python
def calculadora(numero1,numero2,numero3):
  if(numero3==1):
    return(numero1+numero2)
  if(numero3==2):
   return(numero1-numero2)
  if(numero3==3):
    return(numero1*numero2)
  if(numero3==4):
    return(numero1/numero2)

resultado=calculadora(7,10,4)
print("resultado=",resultado)

```
